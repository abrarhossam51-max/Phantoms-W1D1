```python
def process_team_scores(scores_list):

    total_score = 0
    passed_members = []

    #الlist فيها 5 لو مشلنا ال-1 هياخد منهم 4 بس

    for i in range(len(scores_list)):
        score = scores_list[i]
        total_score += score

        #درجه النجاح من 50 يعني نحط >= مش > بس
        if score >= 50:
            passed_members.append(score)

    average = total_score / len(scores_list)

    #المشكله ان total_score مفيهوش مجموع كل الدرجات بسبب الـBug الأول

    return average, passed_members


sample_scores = [45, 50, 75, 90, 60]
avg, passed = process_team_scores(sample_scores)

print(f"Average: {avg}, Passed: {passed}")
```
