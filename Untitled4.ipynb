```python
def student_score(studentgrades):
    total_score = 0
    vailed_count = 0

    for student_grades in studentgrades.values():
        try:
            total_score += float(student_grades)
            vailed_count += 1
        except(TypeError, ValueError):
            continue


    if vailed_count == 0:
            return 0
    return total_score / vailed_count

student_grades ={"Hana": 90,  "Ahmed": 85, "Menna": "Absent", "Ali": 70}


average = student_score(student_grades)
print(average)
```
