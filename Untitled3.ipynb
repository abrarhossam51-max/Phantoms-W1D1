```python
# divide() is responsible only for the division operation
def divide(num1, num2):
    # Raise an error if the second number is zero
    if num2 == 0:
        raise ZeroDivisionError("Cannot divide by zero.")

    return num1 / num2


# read_number() is responsible for getting a valid number from the user
def read_number():
    while True:
        try:
            # Convert the user's input to a float
            return float(input("Enter a number: "))

        # Handle invalid input, such as letters instead of numbers
        except ValueError:
            print("Please enter numbers only.")


# main() organizes the main flow of the program
def main():
    # Get the two numbers from the user
    num1 = read_number()
    num2 = read_number()

    try:
        # Divide the two numbers
        result = divide(num1, num2)
        print("Result:", result)

    # Handle division by zero
    except ZeroDivisionError:
        print("Cannot divide by zero.")


# Start the program
main()
```
